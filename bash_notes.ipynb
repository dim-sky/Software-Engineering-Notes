## 🐧 Running Multiple Commands in Linux

You can chain several commands on one line. The **only difference** between `;` and `&&` is what happens **when the first command fails**.

### 1. `;` — Always continue

```bash
mkdir test ; cd test
```

The second command runs **no matter what**, whether the first succeeded or failed.

### 2. `&&` — Continue only on success

```bash
mkdir test && cd test
```

The second command runs **only if the first one succeeded**.

### ⚖️ Side by side

| | First command **succeeds** | First command **fails** |
|---|---|---|
| `A ; B` | ✅ B runs | ✅ B **still runs** |
| `A && B` | ✅ B runs | ❌ B is **skipped** |

<div style="background: rgba(230, 160, 40, 0.08); border-left: 3px solid #c9973a; padding: 8px 12px; border-radius: 4px; font-size: 0.92em;">
<b>⚠️ Why it matters</b><br>
<code>cd old_logs ; rm *.log</code><br>
If <code>old_logs</code> doesn't exist, <code>cd</code> fails, but <code>rm</code> <b>still runs</b> in your <b>current folder</b> and deletes the wrong files.<br><br>
<code>cd old_logs &amp;&amp; rm *.log</code><br>
Here <code>cd</code> fails, so <code>rm</code> is <b>skipped</b>. Nothing is lost. ✅
</div>

<b>🧠 Easy way to remember</b><br>
<code>;</code> → "do this, <b>then</b> do that"<br>
<code>&amp;&amp;</code> → "do this, and <b>if it worked</b>, do that"
