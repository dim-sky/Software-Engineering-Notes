<big>**Ένα token τη φορά**</big>  
Τα LLM στη βάση τους παράγουν σε κάθε βήμα **μόνο το επόμενο token** (όχι απαραίτητα ολόκληρη λέξη).  
Ένα token μπορεί να είναι μια λέξη, κομμάτι λέξης ή ένα σύμβολο.  
Υπάρχουν και μοντέλα που προβλέπουν πολλά tokens μαζί (π.χ. 2 ή 8), αλλά τα περισσότερα παράγουν **ένα**.

<big>**3 προσεγγίσεις Tokenization**</big>

1. **Character level** — κάθε χαρακτήρας είναι token  
   ✅ Μικρό λεξιλόγιο, δεν υπάρχουν άγνωστες λέξεις  
   ❌ Πολύ μεγάλες ακολουθίες, πιο αργό

2. **Subword level** — κομμάτια λέξεων (π.χ. "παίζ" + "ουμε")  
   ✅ Ισορροπία ανάμεσα στα δύο άκρα  
   ⭐ **Αυτή χρησιμοποιείται κατά κύριο λόγο**

3. **Word level** — κάθε λέξη είναι token  
   ✅ Σύντομες ακολουθίες  
   ❌ Τεράστιο λεξιλόγιο, δεν χειρίζεται άγνωστες λέξεις

<big><big>**Agent vs Deep Agent**</big></big>

<big>**Agent**</big>  
Ένα LLM που σε κάθε βήμα αποφασίζει ποιο εργαλείο (tool) να καλέσει, βλέπει το αποτέλεσμα και συνεχίζει μέχρι να δώσει απάντηση.  
Καλό για **απλές και σύντομες** εργασίες (π.χ. "βρες τον καιρό", "κάνε έναν υπολογισμό").  
Σε μεγάλες εργασίες συχνά χάνεται, ξεχνάει τον στόχο ή σταματάει νωρίς.

<big>**Deep Agent**</big>  
Ένας agent με επιπλέον "εξοπλισμό" για **μεγάλες και σύνθετες** εργασίες:

- **Σχεδιασμός:** φτιάχνει πρώτα λίστα με βήματα (to-do) και την ακολουθεί
- **Sub-agents:** μοιράζει κομμάτια της δουλειάς σε βοηθητικούς agents
- **Αρχεία (filesystem):** αποθηκεύει σημειώσεις και ενδιάμεσα αποτελέσματα, για να μη "γεμίζει" η μνήμη του
- **Αναλυτικές οδηγίες:** έχει μεγάλο και λεπτομερές system prompt

<big>**Με μία φράση**</big>  
Ο agent **εκτελεί**, ο deep agent **οργανώνει, μοιράζει και εκτελεί**.

<big><big>**Τι είναι το Harness**</big></big>

<big>**Με μία φράση**</big>  
Το **harness** είναι ο "σκελετός" γύρω από το LLM: όλα όσα **δεν** είναι το ίδιο το μοντέλο, αλλά του επιτρέπουν να δουλέψει σαν agent.

<big>**Τι περιλαμβάνει**</big>

- **Loop:** καλεί το LLM ξανά και ξανά μέχρι να τελειώσει η εργασία
- **Tools:** του δίνει εργαλεία (αναζήτηση, κώδικας, αρχεία κ.λπ.) και εκτελεί τις κλήσεις τους
- **Prompts:** οι οδηγίες (system prompt) που του λένε πώς να συμπεριφέρεται
- **Μνήμη / context:** τι πληροφορίες βλέπει το μοντέλο σε κάθε βήμα
- **Κανόνες ελέγχου:** πότε σταματάει, τι γίνεται αν κάτι αποτύχει

<big>**Παρομοίωση**</big>  
Το LLM είναι η **μηχανή**, το harness είναι το **υπόλοιπο αυτοκίνητο** (τιμόνι, ρόδες, φρένα).  
Η ίδια μηχανή σε διαφορετικό αυτοκίνητο έχει πολύ διαφορετική απόδοση.

<big>**Παράδειγμα**</big>  
Τα **deep agents** είναι ένα έτοιμο harness: προσθέτουν σχεδιασμό, sub-agents και αρχεία γύρω από το LLM.
