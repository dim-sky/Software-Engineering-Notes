# <span style="color:#1E88E5">⚡ uv & Python Projects – Δομή, εκτέλεση και `__name__`</span>

---

### <span style="color:#1E88E5">❓ Τι είναι το uv;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το **uv** είναι package & project manager για Python, γραμμένο σε Rust (από την Astral). Αντικαθιστά πολλά εργαλεία μαζί και είναι πολύ πιο γρήγορο από το `pip`.

</div>

Γιατί είναι γρήγορο:
- **<span style="color:#43A047">Rust</span>**: γρήγορη μεταγλωττισμένη γλώσσα.
- **<span style="color:#43A047">Παράλληλα downloads</span>**: κατεβάζει πολλά πακέτα ταυτόχρονα.
- **<span style="color:#43A047">Global cache</span>**: κάθε πακέτο κατεβαίνει μία φορά και συνδέεται (hardlink) σε κάθε `.venv`.

| **Παλιό εργαλείο** | **Στο uv** |
|---|---|
| **`pip`** | `uv add` / `uv pip` |
| **`venv` / `virtualenv`** | αυτόματο `.venv` |
| **`pyenv`** | `uv python install` |
| **`pipx`** | `uv tool install` / `uvx` |
| **`poetry` / `pip-tools`** | `uv.lock` |

---

### <span style="color:#1E88E5">❓ Ποια αρχεία έχει ένα uv project;</span>

- **<span style="color:#43A047">`pyproject.toml`</span>**: η ταυτότητα του project (όνομα, έκδοση, dependencies με εύρος εκδόσεων). Ορίζει το **project**.
- **<span style="color:#43A047">`uv.lock`</span>**: οι **ακριβείς** εκδόσεις και hashes όλων των πακέτων και υπο-εξαρτήσεων.
- **<span style="color:#43A047">`.python-version`</span>**: η έκδοση Python του project.
- **<span style="color:#43A047">`.venv/`</span>**: το virtual environment (απομονωμένο περιβάλλον με τα πακέτα του project).

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Αναλογία:** `pyproject.toml` = «τι θέλω» · `uv.lock` = «τι ακριβώς εγκαταστάθηκε». Στο git ανεβαίνουν τα τρία πρώτα, **όχι** το `.venv/`.

</div>

---

### <span style="color:#1E88E5">❓ Πώς δημιουργώ ένα project;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Με **`uv init`**. Τα `--package`, `--lib`, `--script` είναι flags του ίδιου εντολής.

</div>

```bash
uv init                   # project στον τρέχοντα φάκελο
uv init myproject         # δημιουργεί και τον φάκελο
uv init --package myapp   # installable project με δομή src/ (υποστηρίζει [project.scripts])
uv init --lib mylib       # βιβλιοθήκη για άλλους
uv init --script s.py     # μεμονωμένο script με δικά του dependencies
```

```
myapp/                    # δομή με --package
├── pyproject.toml
└── src/
    └── myapp/
        └── __init__.py
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ **Δεν υπάρχει `uv new`.** Το `.venv` και το `uv.lock` δεν τα φτιάχνει το `uv init`, αλλά το πρώτο `uv add` / `uv run` / `uv sync`.

</div>

---

### <span style="color:#1E88E5">❓ Πώς προσθέτω και αφαιρώ πακέτα;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το **`uv add`** γράφει στο `pyproject.toml`, ενημερώνει το `uv.lock` και εγκαθιστά στο `.venv`, όλα μαζί.

</div>

```bash
uv add requests                 # κανονική εξάρτηση
uv add "requests>=2.31"         # με περιορισμό έκδοσης
uv add --dev pytest             # μόνο για development
uv remove requests              # αφαίρεση
uv lock --upgrade               # αναβάθμιση όλων (το lock δεν αλλάζει μόνο του)
uv tree                         # δέντρο εξαρτήσεων
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Μην ανακατεύεις `uv add` με `uv pip install` στο ίδιο project. Το `uv pip` **δεν** ενημερώνει `pyproject.toml` / `uv.lock`.

</div>

### <span style="color:#1E88E5">❓ Τι κάνει το `uv sync` (π.χ. μετά από git clone);</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Στήνει το `.venv` ώστε να ταιριάζει **ακριβώς** με το **`uv.lock`**.

</div>

```
Python λείπει; --εγκατάσταση-->  .venv λείπει; --δημιουργία-->  εγκατάσταση από uv.lock  -->  αφαίρεση ό,τι περισσεύει
```

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 Το `uv run` κάνει **αυτόματα sync**. Μετά από clone αρκεί `uv run ...`.

</div>

### <span style="color:#1E88E5">❓ Τι είναι module, package και `__init__.py`;</span>

- **<span style="color:#43A047">Module</span>**: ένα αρχείο `.py`.
- **<span style="color:#43A047">Package</span>**: φάκελος με modules και `__init__.py`. Γίνεται `import`.
- **<span style="color:#43A047">`__init__.py`</span>**: κάνει τον φάκελο package· εκτελείται **αυτόματα** όταν γίνεται import το package.

```python
# src/myapp/__init__.py
from .core import calculate      # εκθέτει το calculate ως δημόσιο API
__all__ = ["calculate"]          # τι φέρνει το "from myapp import *"
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Το `__init__.py` ορίζει **package**, όχι project (αυτό το κάνει το `pyproject.toml`). Στα μεγάλα projects περιέχει **imports**, όχι «υπογραφές» συναρτήσεων. Το `import myapp` **είναι** ήδη το `__init__.py`· δεν γράφεις ποτέ `myapp.__init__`.

</div>

---

### <span style="color:#1E88E5">❓ Τι σημαίνουν τα `_` και `__` στα ονόματα;</span>

| **Μορφή** | **Σημασία** | **Παράδειγμα** |
|---|---|---|
| **`__όνομα__`** | dunder: δεσμευμένο από την Python, το βάζει ή το καλεί η ίδια | `__name__`, `__init__`, `__len__` |
| **`__όνομα`** (σε κλάση) | name mangling: μετονομάζεται σε `_Κλάση__όνομα` | `self.__secret` |
| **`_όνομα`** | «εσωτερικό», μόνο σύμβαση | `def _helper():` |
| **`όνομα_`** | αποφυγή δεσμευμένης λέξης | `class_` |
| **`_`** | μεταβλητή που δεν μας ενδιαφέρει | `for _ in range(3):` |

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Το `_` **δεν** σημαίνει in-place αλλαγή στην Python. Αυτό ισχύει μόνο ως σύμβαση του **PyTorch** (`x.add_(1)`). Στην Python: `lista.sort()` (in-place) vs `sorted(lista)` (νέα λίστα).

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι τα `__name__` και `__package__`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Είναι μεταβλητές που η Python βάζει **σε κάθε module** όταν το φορτώνει. Δεν υπάρχει ένα κοινό global object· κάθε module έχει τις δικές του.

</div>

- **<span style="color:#43A047">`__name__`</span>**: `"__main__"` αν το αρχείο **εκτελείται**, αλλιώς το πλήρες όνομα του module (π.χ. `myapp.data.loader`).
- **<span style="color:#43A047">`__package__`</span>**: το **γονικό** package (π.χ. `myapp.data`) ή `None` αν η Python δεν ξέρει σε ποιο package ανήκει.

```python
for k, v in globals().items():      # όλα τα ονόματα του τρέχοντος module
    if k.startswith("__"):          # μόνο τα dunder
        print(k, "=", v)            # __name__, __package__, __file__, ...

import sys
print(sys.modules["myapp"].__file__)  # τα module objects όλων των φορτωμένων modules
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Το uv **δεν** φτιάχνει κανένα global object. Ετοιμάζει το `.venv` και ξεκινά την Python· μετά ο κώδικας τρέχει ως κανονική Python. Για prints χρησιμοποίησε `f"..."`, όχι `t"..."` (template string, Python 3.14).

</div>

---

### <span style="color:#1E88E5">❓ Τι κάνει το `if __name__ == "__main__":`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Ρωτάει «είμαι το αρχείο που **ξεκίνησε** το πρόγραμμα;». Ο κώδικας από κάτω τρέχει **μόνο** τότε, όχι όταν το αρχείο γίνεται **import**.

</div>

```python
def main():                       # ο εκτελέσιμος κώδικας σε συνάρτηση
    print("Τρέχω")

if __name__ == "__main__":        # μόνο όταν εκτελείται απευθείας
    main()
```

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 Δεν απαιτεί `main()`, δουλεύει με οποιονδήποτε κώδικα. Σύμβαση όμως: μέσα βάζεις **μόνο** `main()` (οι μεταβλητές δεν γίνονται global, η `main()` χρησιμοποιείται σε entry points και tests).

</div>

---

### <span style="color:#1E88E5">❓ Με ποιους τρόπους τρέχω κώδικα και σε τι διαφέρουν;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το `uv run` κάνει πάντα sync + εκτέλεση στο `.venv`. Η πραγματική διαφορά είναι **τι** ακολουθεί: **path**, **`-m`** ή **εντολή**.

</div>

| **Κριτήριο** | **`uv run src/myapp/data/loader.py`** | **`uv run python -m myapp.data.loader`** | **`uv run myapp-load`** ([project.scripts]) |
|---|---|---|---|
| **Στόχος** | path με `/` και `.py` | όνομα import με `.` | όνομα εντολής |
| **`__name__`** | `__main__` | `__main__` | `myapp.data.loader` |
| **`__package__`** | `None` | `myapp.data` | `myapp.data` |
| **Τρέχει το `if __name__`** | ✅ | ✅ | ❌ |
| **Τρέχει το `__init__.py`** | ❌ | ✅ | ✅ |
| **Relative imports** | ❌ σφάλμα | ✅ | ✅ |

```bash
uv run python -m myapp.data.loader    # σωστό
uv run -m myapp.data.loader           # συντόμευση του παραπάνω
uv run python -m myapp                # εκτελεί το myapp/__main__.py
uv run python -m http.server          # τρέχει και modules της standard library
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Στο `-m`: **όχι** `src.`, **όχι** `.py`, **όχι** `:συνάρτηση`, **όχι** `/` ή `:` αντί για τελείες. Το `:` υπάρχει **μόνο** στο `[project.scripts]`.

</div>

---

### <span style="color:#1E88E5">❓ Τι συμβαίνει βήμα-βήμα με `uv run python -m myapp.data.loader`;</span>

```
uv: sync .venv  -->  Python ξεκινά  --import-->  myapp/__init__.py  --import-->  myapp/data/__init__.py  --εκτέλεση ως __main__-->  loader.py
```

- Η Python **δεν** «χτίζει» όλο το project. Φορτώνει μόνο την **αλυσίδα** που χρειάζεται: γονικά packages, στόχο, και ό,τι κάνουν αυτά import.
- Ο στόχος **δεν** γίνεται κανονικό import· εκτελείται με `__name__ = "__main__"`.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Αν κάποιο άλλο module κάνει μετά `import myapp.data.loader`, το αρχείο φορτώνεται **ξανά** ως ξεχωριστό module (ο top-level κώδικας τρέχει δύο φορές). Γι' αυτό εκτελείς entry points, όχι modules βιβλιοθήκης.

</div>

---

### <span style="color:#1E88E5">❓ Τι πρέπει να ισχύει για να τρέξει κάτι με `python -m`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Ό,τι χρειάζεται και ένα κανονικό **`import`**, συν `__main__.py` αν ο στόχος είναι package.

</div>

| **Προϋπόθεση** | **Απαραίτητο;** |
|---|---|
| **Βρίσκεται στον τρέχοντα φάκελο ή είναι εγκατεστημένο στο `.venv`** | ✅ (με `--package` το project εγκαθίσταται editable) |
| **Όνομα με τελείες, χωρίς `.py`, χωρίς `src`** | ✅ |
| **`__init__.py` σε κάθε φάκελο της διαδρομής** | ✅ σύμβαση |
| **`__main__.py` αν ο στόχος είναι package** | ✅ |
| **`main()` / `if __name__`** | ❌ |

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Μην δίνεις ονόματα της standard library (`test`, `random`, `json`): μπορεί να φορτωθεί το λάθος module. Χρησιμοποίησε `snake_case`, όχι παύλες (`data-loader.py` δεν γίνεται import).

</div>

### <span style="color:#1E88E5">❓ Πώς δουλεύει το `[project.scripts]`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Ορίζει μια **εντολή** που κάνει **import** ένα module και **καλεί μια συνάρτηση** χωρίς ορίσματα. Απαιτεί installable project (`uv init --package`).

</div>

```toml
[project.scripts]
myapp = "myapp.starthere:main"    # εντολή = "package.module:συνάρτηση"
```

```python
# αυτό κάνει ουσιαστικά ο εκκινητής του uv:
from myapp.starthere import main   # 1. import → τρέχει ΟΛΟ το top-level του module
main()                             # 2. κλήση της συνάρτησης
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Το μέρος μετά το `:` είναι **συνάρτηση**, όχι module. Χωρίς συνάρτηση δεν γίνεται entry point. Top-level κώδικας (π.χ. `myFunction(tempMessage)`) τρέχει **επιπλέον** λόγω του import. Βάζεις **μόνο** entry points, όχι εντολή για κάθε module.

</div>

### <span style="color:#1E88E5">❓ Τι είναι το `__main__.py`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Αρχείο με ειδικό όνομα μέσα σε package: ορίζει τι τρέχει με **`python -m package`**.

</div>

```python
# src/myapp/__main__.py
from myapp.starthere import main   # φέρνει το entry point
main()                             # χωρίς if __name__: τρέχει πάντα ως κύριο πρόγραμμα
```

| **Κριτήριο** | **`[project.scripts]`** | **`__main__.py`** |
|---|---|---|
| **Εντολή** | `uv run myapp` | `uv run python -m myapp` |
| **Χρειάζεται `--package`** | ✅ | ❌ |
| **Τι δείχνει** | `module:συνάρτηση` | εκτελεί το αρχείο |

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 Συνήθως έχεις **και τα δύο**, και οδηγούν στην **ίδια** `main()`. Έτσι δουλεύουν τα `python -m pip`, `python -m venv`.

</div>

### <span style="color:#1E88E5">❓ Τι κώδικας μπαίνει σε κάθε είδος αρχείου;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 **Τα modules ορίζουν πράγματα, τα entry points τα εκτελούν.**

</div>

```python
# src/myapp/data/loader.py  (module βιβλιοθήκης)
import logging                          # imports: ΟΚ στο top-level
MAX_ROWS = 1000                         # σταθερές: ΟΚ
logger = logging.getLogger(__name__)    # δημιουργία logger: ΟΚ (δεν τυπώνει)

def load(path):                         # ορισμοί: ΟΚ
    logger.info("Loading %s", path)     # logs/prints/IO: ΜΟΝΟ μέσα σε συναρτήσεις
```

| **Αρχείο** | **`main()`** | **`if __name__`** | **Πώς τρέχει** |
|---|---|---|---|
| **Module βιβλιοθήκης** | ❌ | ❌ (μόνο για γρήγορη δοκιμή στην εξάσκηση) | γίνεται import |
| **Entry point (`starthere.py`)** | ✅ | ✅ | `uv run myapp` |
| **`__init__.py`** | ❌ | ❌ | αυτόματα στο import |
| **`__main__.py`** | καλεί το `main()` | ❌ | `uv run python -m myapp` |
| **Πρόχειρο script** | προαιρετικό | προαιρετικό | `uv run script.py` |

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Κώδικας κάτω από το `if __name__` **δεν τρέχει ποτέ** στο import. Κώδικας έξω από αυτό τρέχει **σε κάθε** import (side effect). Logs τύπου «το module φορτώθηκε» δεν χρειάζονται.

</div>

### <span style="color:#1E88E5">❓ Τι είναι το `__pycache__`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Φάκελος με **bytecode** (`.pyc`, μεταγλωττισμένη μορφή του κώδικα) για γρηγορότερη φόρτωση. Δημιουργείται **μόνο** σε φακέλους με modules που έγιναν **import**.

</div>

- Ξαναφτιάχνεται αυτόματα όταν αλλάζει το `.py`· διαγράφεται άφοβα.
- Το αρχείο που τρέχεις με **path** δεν μπαίνει σε cache (με `-m` μπαίνει).
- Δεν ανεβαίνει στο git (το εξαιρεί το `.gitignore` του `uv init`).
- Απενεργοποίηση: `PYTHONDONTWRITEBYTECODE=1` ή `python -B`.

### <span style="color:#1E88E5">❓ Πώς τρέχω μεμονωμένα scripts με δικά τους dependencies;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Με **inline metadata** (PEP 723): τα dependencies δηλώνονται μέσα στο ίδιο το αρχείο, χωρίς project.

</div>

```bash
uv init --script fetch.py            # δημιουργεί script με metadata
uv add --script fetch.py requests    # προσθέτει dependency μέσα στο αρχείο
uv run fetch.py                      # φτιάχνει προσωρινό περιβάλλον και τρέχει
uv run --with rich fetch.py          # προσωρινό επιπλέον πακέτο, χωρίς αλλαγή αρχείου
```

```python
# /// script
# requires-python = ">=3.12"
# dependencies = ["requests"]
# ///
import requests                                          # διαθέσιμο χάρη στο metadata
print(requests.get("https://example.com").status_code)  # τυπώνει τον κωδικό HTTP
```

### <span style="color:#1E88E5">❓ Ποια είναι τα best practices;</span>

**Κάνε πάντα:**
- `uv init --package` για οτιδήποτε μεγαλύτερο από script.
- Εκτελέσιμο κώδικα σε `main()` + `if __name__ == "__main__": main()` **μόνο** στα entry points.
- Modules μέσα σε package τα τρέχεις με **`-m`**.
- **Absolute imports** (`from myapp.data.loader import load`), σύσταση PEP 8.
- Commit `pyproject.toml`, `uv.lock`, `.python-version`.

**Μην κάνεις ποτέ:**
- Side effects στο top-level modules βιβλιοθήκης.
- Κώδικα στο `__init__.py` πέρα από imports για το δημόσιο API.
- `sys.path.append(...)` για να «διορθώσεις» imports.
- Τρέξιμο αρχείου μέσα στο `src/` με path.

| **Τρόπος εκτέλεσης** | **Πότε** |
|---|---|
| **`uv run myapp`** | ✅ κύριος τρόπος (`[project.scripts]`) |
| **`uv run python -m myapp`** | ✅ ισοδύναμη εναλλακτική (`__main__.py`) |
| **`uv run python -m myapp.dir.module`** | ✅ γρήγορη δοκιμή module στην εξάσκηση |
| **`uv run pytest`** | ✅ έλεγχος συναρτήσεων σε σοβαρό project |
| **`uv run script.py`** | ✅ μόνο για αρχεία **εκτός** `src/` |
| **`uv run src/myapp/.../module.py`** | ❌ σπάνε τα imports |
| **`python αρχείο.py` χωρίς `uv run`** | ❌ μπορεί να τρέξει εκτός `.venv` |
