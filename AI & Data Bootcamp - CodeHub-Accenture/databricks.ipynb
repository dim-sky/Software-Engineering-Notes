# <span style="color:#1E88E5">🏠 Databricks Lakehouse – Βασική Ορολογία</span>

---

### <span style="color:#1E88E5">❓ Τι είναι το Object Store;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το **Object Store** (αποθήκη αντικειμένων) είναι ιδιαίτερα επεκτάσιμος, ανθεκτικός και φθηνός cloud χώρος αποθήκευσης αρχείων, π.χ. AWS S3.

</div>

- Αποθηκεύει **objects** (αρχεία + metadata) και όχι πίνακες βάσης δεδομένων.
- Είναι το **φυσικό επίπεδο αποθήκευσης** του lakehouse.
- Παραδείγματα: AWS S3, Azure Data Lake Storage (ADLS), Google Cloud Storage.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Το object store από μόνο του **δεν** έχει ACID, schemas ή governance. Αυτά τα προσθέτει ένα επιπλέον επίπεδο (στο Databricks: το **Delta Lake**).

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι το Metadata;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το **Metadata** είναι «δεδομένα για τα δεδομένα»: περιγράφει τα αποθηκευμένα δεδομένα χωρίς να είναι τα ίδια τα δεδομένα.

</div>

- **<span style="color:#43A047">Schema</span>**: στήλες και τύποι δεδομένων.
- **<span style="color:#43A047">Lineage</span>** (γενεαλογία): από πού προήλθε ένα dataset και πώς μετασχηματίστηκε.
- **<span style="color:#43A047">Permissions</span>**: ποιος έχει πρόσβαση σε τι.
- **<span style="color:#43A047">Statistics</span>**: μέγεθος, πλήθος γραμμών, min/max τιμές.

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Αναλογία:** Το metadata είναι η **καρτέλα βιβλίου** στη βιβλιοθήκη: δεν είναι το βιβλίο, αλλά σου λέει τι είναι και πού βρίσκεται.

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι τα ACID Transactions;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Τα **ACID Transactions** εγγυώνται ότι κάθε αλλαγή στα δεδομένα είναι αξιόπιστη, ακόμα και με σφάλματα ή ταυτόχρονους χρήστες.

</div>

- **<span style="color:#43A047">Atomicity</span>** (ατομικότητα): η αλλαγή γίνεται εξ ολοκλήρου ή καθόλου.
- **<span style="color:#43A047">Consistency</span>** (συνέπεια): τα δεδομένα μένουν πάντα έγκυρα, σύμφωνα με τους κανόνες/schema.
- **<span style="color:#43A047">Isolation</span>** (απομόνωση): ταυτόχρονες αλλαγές δεν επηρεάζουν η μία την άλλη.
- **<span style="color:#43A047">Durability</span>** (διάρκεια): μια επιβεβαιωμένη αλλαγή δεν χάνεται.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Τα κλασικά data lakes **δεν** έχουν ACID: ένα job που αποτυγχάνει στη μέση αφήνει μισογραμμένα αρχεία. Αυτό είναι ένας από τους βασικούς λόγους ύπαρξης του lakehouse.

</div>

---

### <span style="color:#1E88E5">❓ Τι διαφέρει το Batch από το Streaming Data;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Τα **Batch data** επεξεργάζονται μαζικά, ανά τακτά διαστήματα. Τα **Streaming data** επεξεργάζονται συνεχώς, μόλις φτάνουν.

</div>

| Κριτήριο | **Batch** | **Streaming** |
|---|---|---|
| **Επεξεργασία** | Σε ομάδες, περιοδικά | Συνεχώς, εγγραφή-εγγραφή |
| **Δεδομένα** | Ιστορικά, μεγάλα σύνολα | Νέα γεγονότα σε πραγματικό χρόνο |
| **Πηγή** | Αρχεία, πίνακες | Message queues (π.χ. Kafka), εξωτερικά events |
| **Καθυστέρηση** | Λεπτά–ώρες | Δευτερόλεπτα |
| **Παράδειγμα** | Ημερήσια αναφορά πωλήσεων | Ανίχνευση απάτης σε πληρωμές |

```python
df = spark.read.format("delta").load("/data/sales")              # Batch: διαβάζει τα υπάρχοντα δεδομένα μία φορά
stream_df = spark.readStream.format("delta").load("/data/sales")  # Streaming: διαβάζει συνεχώς τα νέα δεδομένα
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Τα δύο δεν είναι αντίπαλα. Στο lakehouse **ο ίδιος πίνακας** μπορεί να εξυπηρετεί και batch και streaming, χωρίς ξεχωριστά συστήματα.

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι το ETL;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 **ETL** (Extract, Transform, Load) είναι η διαδικασία μεταφοράς δεδομένων από ένα σύστημα σε άλλο, συχνά με μετατροπή μορφής.

</div>

```
Extract (εξαγωγή)  --καθαρισμός/μετατροπή-->  Transform  --εγγραφή-->  Load (φόρτωση στον προορισμό)
```

- **<span style="color:#43A047">Extract</span>**: διαβάζω τα δεδομένα από την πηγή.
- **<span style="color:#43A047">Transform</span>**: καθαρίζω, μετατρέπω, εμπλουτίζω.
- **<span style="color:#43A047">Load</span>**: γράφω στον προορισμό (warehouse ή lake).

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Υπάρχει και το **ELT**: πρώτα φορτώνω τα ακατέργαστα δεδομένα στον προορισμό και μετασχηματίζω εκεί. Είναι η συνηθισμένη προσέγγιση στα lakes και στα lakehouses.

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι το Data Lake;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το **Data Lake** είναι ένα κεντρικό αποθετήριο που αποθηκεύει **όλους** τους τύπους δεδομένων (structured, semi-structured, unstructured) σε ωμή μορφή.

</div>

- **<span style="color:#43A047">Structured</span>**: πίνακες, CSV.
- **<span style="color:#43A047">Semi-structured</span>**: JSON, XML.
- **<span style="color:#43A047">Unstructured</span>**: εικόνες, βίντεο, κείμενο, logs.
- Ευέλικτα schemas (**schema-on-read**: το schema εφαρμόζεται όταν διαβάζω, όχι όταν γράφω).
- Χτίζεται συνήθως πάνω σε object store, άρα είναι φθηνό.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Χωρίς governance, ACID και έλεγχο ποιότητας, το lake εκφυλίζεται σε **data swamp** (βάλτο δεδομένων): δεδομένα που κανείς δεν εμπιστεύεται.

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι το Data Warehouse;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το **Data Warehouse** είναι σύστημα βασισμένο σε σχεσιακή βάση, βελτιστοποιημένο για ανάλυση και reporting, κυρίως με **structured** δεδομένα.

</div>

- Αυστηρά schemas (**schema-on-write**: το schema ελέγχεται κατά την εγγραφή).
- Ισχυρό governance και ACID.
- Λιγότερο ευέλικτο: δυσκολεύεται με unstructured δεδομένα και ML workloads.
- Συνήθως ακριβότερο, συχνά με κλειστές (proprietary) μορφές αποθήκευσης.

---

### <span style="color:#1E88E5">❓ Data Lake vs Data Warehouse: τι διαφέρει;</span>

| Κριτήριο | **Data Lake** | **Data Warehouse** |
|---|---|---|
| **Τύποι δεδομένων** | Όλοι | Κυρίως structured |
| **Schema** | Ευέλικτο (schema-on-read) | Αυστηρό (schema-on-write) |
| **Governance** | Συχνά ελλιπές | Ισχυρό |
| **ACID** | Συνήθως όχι | Ναι |
| **Κόστος** | Χαμηλό | Υψηλότερο |
| **Κύρια χρήση** | Data science, ML, ωμά δεδομένα | BI, reporting |

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Συμπέρασμα:** Το Lake δίνει **ευελιξία χωρίς έλεγχο**. Το Warehouse δίνει **έλεγχο χωρίς ευελιξία**.

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι το Lakehouse;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το **Lakehouse** συνδυάζει τα πλεονεκτήματα του data lake και του data warehouse: governance, ACID και ευελιξία ανάλυσης πάνω σε **φθηνό object store**.

</div>

```
Object Store (S3)  -->  + Delta Lake (ACID, schema, metadata)  -->  Batch + Streaming + SQL + ML στο ίδιο σύστημα
```

- **<span style="color:#43A047">Object store</span>**: φθηνή και επεκτάσιμη αποθήκευση.
- **<span style="color:#43A047">Metadata enforcement</span>**: schema και governance επιβάλλονται πάνω στα αρχεία.
- **<span style="color:#43A047">ACID</span>**: αξιόπιστες αλλαγές στους πίνακες.
- **<span style="color:#43A047">Streaming pipelines</span>**: ζωντανά δεδομένα στους ίδιους πίνακες με τα batch.

```python
df.write.format("delta").mode("append").save("/data/sales")  # Γράφει με ACID εγγυήσεις πάνω στο object store
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Το lakehouse **δεν** είναι δύο συστήματα κολλημένα μαζί (lake + warehouse με ETL ανάμεσα). Είναι **ένα** σύστημα με ένα κοινό αντίγραφο δεδομένων.

</div>

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Αναλογία:** Lake = αποθήκη με ό,τι να ’ναι. Warehouse = τακτοποιημένο ράφι σούπερ μάρκετ. Lakehouse = η αποθήκη με ράφια και ταμείο μέσα της.

</div>
