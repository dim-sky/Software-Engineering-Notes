# <span style="color:#1E88E5">🐳 Docker – Βασικές Έννοιες</span>

---

### <span style="color:#1E88E5">❓ Τι είναι το Docker;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 **Πλατφόρμα για τη δημιουργία, εκτέλεση και διανομή containers.**

</div>

Αποτελείται από:
- **<span style="color:#43A047">Docker Engine</span>** (`dockerd`): τρέχει τα containers.
- **<span style="color:#43A047">Docker CLI</span>** (`docker`): η εντολή με την οποία δίνουμε οδηγίες στο Engine.
- **<span style="color:#43A047">Registries</span>** (π.χ. Docker Hub): αποθήκες από όπου κατεβάζουμε (`pull`) ή ανεβάζουμε (`push`) images.

---

### <span style="color:#1E88E5">❓ Τι είναι ένα container;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Μια **απομονωμένη διεργασία (process)** που τρέχει στον host μαζί με ό,τι χρειάζεται η εφαρμογή.

</div>

- **Δεν έχει δικό του kernel**: μοιράζεται τον kernel του host. Γι' αυτό είναι πολύ πιο ελαφρύ από ένα VM.
- **Έχει όμως δικό του σύστημα αρχείων**, που συνήθως περιέχει βασικά αρχεία μιας διανομής Linux (π.χ. Debian, Alpine).
- Η απομόνωση γίνεται μέσω του Linux kernel:
  - **<span style="color:#43A047">namespaces</span>**: τι "βλέπει" το container (αρχεία, δίκτυο, διεργασίες).
  - **<span style="color:#43A047">cgroups</span>**: πόσους πόρους μπορεί να χρησιμοποιήσει (CPU, RAM).

---

### <span style="color:#1E88E5">❓ Ποια η διαφορά container και VM;</span>

| | VM | Container |
|---|---|---|
| **Kernel** | Δικός του | Του host |
| **Μέγεθος** | GB | MB |
| **Εκκίνηση** | Λεπτά | Δευτερόλεπτα |
| **Απομόνωση** | Πλήρης (virtualized hardware) | Σε επίπεδο διεργασίας |

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Σε **Windows/Mac**, το Docker Desktop τρέχει στο παρασκήνιο μια μικρή Linux VM, γιατί τα Linux containers χρειάζονται Linux kernel.

</div>

---

### <span style="color:#1E88E5">❓ Τι γίνεται με τα δεδομένα ενός container;</span>

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Τα containers είναι **προσωρινά (ephemeral)**: αν διαγραφεί το container, χάνονται όσα έγραψε μέσα του.

</div>

Για μόνιμα δεδομένα (π.χ. βάση δεδομένων) χρησιμοποιούμε **<span style="color:#43A047">volumes</span>**, δηλαδή αποθηκευτικό χώρο που ζει έξω από το container.

---

### <span style="color:#1E88E5">❓ Τι είναι ένα image;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Ένα **read-only πρότυπο** που περιέχει ό,τι χρειάζεται για να τρέξει η εφαρμογή.

</div>

Περιέχει:
- βασικό σύστημα αρχείων (π.χ. Alpine),
- runtime (π.χ. Node.js, Python),
- εξαρτήσεις (βιβλιοθήκες),
- τον κώδικα,
- την εντολή εκκίνησης.

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Αναλογία:** image = κλάση, container = αντικείμενο. Από ένα image τρέχουμε όσα containers θέλουμε.

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι τα layers ενός image;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Κάθε layer είναι μια **αλλαγή στο σύστημα αρχείων** που προκύπτει από μια εντολή του Dockerfile (κυρίως `FROM`, `RUN`, `COPY`). Τα layers στοιβάζονται το ένα πάνω στο άλλο.

</div>

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Τα layers **δεν** είναι διαφορετικά projects (frontend, backend κ.λπ.).

</div>

Τα layers **γίνονται cache**: αν αλλάξει ένα layer, ξαναχτίζονται μόνο αυτό και όσα ακολουθούν. Γι' αυτό βάζουμε **πρώτα** ό,τι αλλάζει σπάνια (dependencies) και **τελευταίο** ό,τι αλλάζει συχνά (κώδικας).

---

### <span style="color:#1E88E5">❓ Πώς οργανώνω μια εφαρμογή με frontend, backend και βάση;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 **Μία υπηρεσία ανά container.** Frontend, backend και βάση έχουν το καθένα δικό του image και container. Για να τα τρέξουμε μαζί χρησιμοποιούμε το **Docker Compose**.

</div>

---

### <span style="color:#1E88E5">❓ Τι είναι το Dockerfile;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Αρχείο με οδηγίες για το πώς **χτίζεται ένα image** (όχι το container).

</div>

```
Dockerfile  --docker build-->  Image  --docker run-->  Container
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Ρυθμίσεις εκτέλεσης (θύρες, volumes, μεταβλητές περιβάλλοντος) δίνονται συνήθως κατά το `docker run` ή στο Compose, όχι στο Dockerfile.

</div>

**Παράδειγμα (Node.js):**
```dockerfile
FROM node:20-alpine          # base image
WORKDIR /app                 # φάκελος εργασίας μέσα στο image
COPY package*.json ./        # πρώτα τα dependencies (caching)
RUN npm install
COPY . .                     # μετά ο κώδικας
EXPOSE 3000                  # τεκμηρίωση της θύρας της εφαρμογής
CMD ["node", "server.js"]    # εντολή εκκίνησης του container
```

---

### <span style="color:#1E88E5">❓ Ποιες είναι οι βασικές εντολές;</span>

```bash
docker build -t my-app .            # χτίζει image με όνομα my-app από το Dockerfile του τρέχοντος φακέλου
docker run -d -p 8080:3000 my-app   # τρέχει container στο παρασκήνιο (-d), θύρα 8080 host -> 3000 container
docker ps                           # λίστα με τα containers που τρέχουν
docker logs <id>                    # έξοδος της εφαρμογής
docker stop <id>                    # σταματά το container
```

---

### <span style="color:#1E88E5">❓ Τι είναι το Docker Compose;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Εργαλείο για να ορίζουμε και να τρέχουμε **πολλά containers μαζί** μέσω ενός αρχείου `compose.yaml`.

</div>

Τα containers μπαίνουν σε κοινό δίκτυο και βρίσκουν το ένα το άλλο με το **όνομα της υπηρεσίας** (π.χ. το backend συνδέεται στη βάση ως `db`).

```yaml
services:
  backend:
    build: ./backend            # χτίζεται από το Dockerfile του backend
    ports:
      - "8080:3000"
    environment:
      DB_HOST: db
    depends_on:
      - db
  db:
    image: postgres:16          # έτοιμο image από το Docker Hub
    environment:
      POSTGRES_PASSWORD: secret
    volumes:
      - dbdata:/var/lib/postgresql/data   # μόνιμα δεδομένα

volumes:
  dbdata:
```

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

▶️ Εκκίνηση όλων: `docker compose up`

</div>

# <span style="color:#1E88E5">🐳 Docker – Dockerfile για FastAPI με uv: Λογική και Επιλογές</span>

---

## <span style="color:#1E88E5">🧭 Γενική λογική</span>

### <span style="color:#1E88E5">❓ Πώς σκέφτομαι όταν γράφω ένα Dockerfile;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το Dockerfile είναι τα **βήματα που θα έκανα για να στήσω την εφαρμογή σε ένα καινούριο, άδειο μηχάνημα**, γραμμένα με τη σειρά. Κάθε γραμμή απαντά σε μια ερώτηση («από πού ξεκινάω;», «τι εγκαθιστώ;», «τι αντιγράφω;»), και συχνά υπάρχουν **πολλές σωστές απαντήσεις**.

</div>

Για μια FastAPI εφαρμογή σε καθαρό μηχάνημα θα έκανα:
1. Εγκατάσταση Python
2. Εγκατάσταση uv
3. Εγκατάσταση των εξαρτήσεων
4. Αντιγραφή του κώδικα
5. Εκκίνηση του server (`uvicorn`)

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Κανόνας επιλογών:** ξεκινάω από την απλούστερη επιλογή που δουλεύει σωστά, και αλλάζω μόνο όταν έχω συγκεκριμένο λόγο (μέγεθος, ταχύτητα build, ασφάλεια).

</div>

---

### <span style="color:#1E88E5">❓ Ποια είναι η δομή του project;</span>

```
my-api/
├── app/
│   └── main.py          # ο κώδικας
├── pyproject.toml       # τι εξαρτήσεις χρειάζομαι (με εύρος εκδόσεων)
├── uv.lock              # ποια ακριβώς έκδοση της καθεμίας
├── .dockerignore        # τι ΔΕΝ μπαίνει στο image
└── Dockerfile
```

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το `pyproject.toml` λέει **τι** χρειάζομαι (π.χ. `fastapi>=0.115`). Το `uv.lock` λέει **ποια ακριβώς έκδοση** εγκαταστάθηκε. Χάρη στο lock, το image έχει ακριβώς ό,τι δοκίμασα τοπικά.

</div>

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Το `uvicorn` πρέπει να είναι στις εξαρτήσεις (`uv add uvicorn` ή `uv add "fastapi[standard]"`). Αλλιώς το `CMD` δεν θα το βρει.

</div>

---

### <span style="color:#1E88E5">❓ Πώς είναι ολόκληρο το Dockerfile;</span>

```dockerfile
FROM python:3.12-slim

COPY --from=ghcr.io/astral-sh/uv:X.Y.Z /uv /uvx /bin/

ENV PYTHONDONTWRITEBYTECODE=1 \
    PYTHONUNBUFFERED=1

WORKDIR /app

COPY pyproject.toml uv.lock ./
RUN uv sync --locked --no-install-project --no-dev --no-cache

COPY ./app ./app
RUN uv sync --locked --no-dev --no-cache

ENV PATH="/app/.venv/bin:$PATH"

EXPOSE 8000

CMD ["uvicorn", "app.main:app", "--host", "0.0.0.0", "--port", "8000"]
```

Παρακάτω εξηγείται κάθε βήμα.

---

## <span style="color:#1E88E5">📦 FROM: Base image</span>

### <span style="color:#1E88E5">❓ Τι σημαίνει `FROM python:3.12-slim`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 «Ξεκίνα από αυτό το **έτοιμο image**». Δεν χτίζω ποτέ από το μηδέν: παίρνω ένα υπάρχον image και προσθέτω από πάνω τα δικά μου layers.

</div>

```
FROM   python   :   3.12-slim
 │       │              │
 │       │              └── tag: έκδοση και παραλλαγή
 │       └── όνομα image
 └── "ξεκίνα από εδώ"
```

- **<span style="color:#43A047">python</span>**: επίσημο image από το Docker Hub. Ένα μικρό Linux (Debian) με ήδη εγκατεστημένη Python.
- **<span style="color:#43A047">3.12</span>**: ίδια έκδοση με την τοπική μου (ή με το `requires-python` του `pyproject.toml`).
- **<span style="color:#43A047">slim</span>**: μόνο τα απαραίτητα για να τρέξει Python.

Στο `docker build`, αν το image δεν υπάρχει τοπικά, κατεβαίνει από το Docker Hub (μόνο την πρώτη φορά).

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

▶️ Εξερεύνηση ενός image από μέσα: `docker run -it --rm python:3.12-slim bash`

</div>

---

### <span style="color:#1E88E5">❓ Πώς επιλέγω base image για οποιαδήποτε τεχνολογία;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Τρεις ερωτήσεις με τη σειρά:
1. **Τι χρειάζεται για να τρέξει;** → Το επίσημο image της γλώσσας.
2. **Ποια έκδοση;** → Ίδια με την τοπική. Ποτέ `latest`.
3. **Ποια παραλλαγή;** → `slim`, εκτός αν έχω λόγο για κάτι άλλο.

</div>

| Εφαρμογή | Base image |
|---|---|
| **Python** | `python:3.12-slim` |
| **Node.js** | `node:22-slim` |
| **Java** | `eclipse-temurin:21-jre` |
| **Go** | `golang:1.23` (για build) |
| **Στατικό frontend** | `nginx:alpine` |

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Προτιμώ images με ένδειξη **Docker Official Image** ή **Verified Publisher** στο Docker Hub. Images τυχαίων χρηστών μπορεί να έχουν οτιδήποτε μέσα ή να μην ενημερώνονται.

</div>

---

### <span style="color:#1E88E5">❓ `slim`, `alpine` ή πλήρες;</span>

Η βασική διαφορά είναι η **βιβλιοθήκη C** του συστήματος: το Debian (`slim`) έχει **glibc**, το Alpine έχει **musl**. Πακέτα με compiled κώδικα (C/Rust) διανέμονται ως έτοιμα binaries (**wheels**) φτιαγμένα για συγκεκριμένη βιβλιοθήκη C.

| Επιλογή | Μέγεθος βάσης | Πλεονέκτημα | Μειονέκτημα | Πότε |
|---|---|---|---|---|
| **`slim`** | ~150 MB | Δουλεύει με όλα τα πακέτα | Λίγο μεγαλύτερο | **Προεπιλογή** |
| **`alpine`** | ~50 MB | Πολύ μικρό | Αν λείπει wheel για musl → compile: αργό build, μεγαλύτερο image | Όταν το μέγεθος μετράει και οι εξαρτήσεις δεν έχουν πρόβλημα |
| **πλήρες** (`python:3.12`) | ~1 GB | Έχει compilers και εργαλεία | Πολύ μεγάλο | Όταν πρέπει να κάνω compile |

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Τεστ για alpine:** τρέχω `docker build`. Αν περάσει χωρίς να αρχίσει να κάνει compile πακέτα, είμαι εντάξει. (Στο Alpine: `apk` αντί για `apt`, `sh` αντί για `bash`.)

</div>

---

### <span style="color:#1E88E5">❓ Γιατί κάποια Dockerfiles έχουν πολλά `FROM`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Σε κάποιες τεχνολογίες **άλλο χρειάζομαι για να χτίσω και άλλο για να τρέξω**. Με **multi-stage build** χτίζω σε ένα μεγάλο image και αντιγράφω μόνο το αποτέλεσμα σε ένα μικρό (με `COPY --from`).

</div>

- **Python, Node**: ο κώδικας τρέχει όπως είναι → συνήθως αρκεί ένα image.
- **Go, Java**: compile → το toolchain χρειάζεται μόνο για το build.
- **React frontend**: Node για το `npm run build`, nginx για να σερβίρει τα αρχεία που προκύπτουν.

## <span style="color:#1E88E5">📂 Η εντολή COPY</span>

### <span style="color:#1E88E5">❓ Πώς συντάσσεται το `COPY`;</span>

```dockerfile
COPY [επιλογές] <πηγή>... <προορισμός>
```

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 **Το τελευταίο όρισμα είναι πάντα ο προορισμός. Όλα τα προηγούμενα είναι πηγές.** Όσα ξεκινούν με `--` είναι επιλογές. Ίδια λογική με το `cp` του Linux.

</div>

- **<span style="color:#43A047">Πηγή</span>**: path μέσα στο **build context** (ο φάκελος που δίνω στο `docker build`, π.χ. το `.`). Δεν μπορεί να βγει έξω από αυτόν (`../` δεν δουλεύει). Ό,τι είναι στο `.dockerignore` αγνοείται. Επιτρέπονται wildcards (`*.json`).
- **<span style="color:#43A047">Προορισμός</span>**: path μέσα στο image. Αν είναι σχετικό, ξεκινά από το `WORKDIR`. Αν δεν υπάρχει, δημιουργείται.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Με **πολλές πηγές**, ο προορισμός **πρέπει να τελειώνει σε `/`** (π.χ. `./`), για να είναι σαφές ότι είναι φάκελος. Αλλιώς το build βγάζει σφάλμα.

</div>

---

### <span style="color:#1E88E5">❓ Μπορεί η πηγή να είναι φάκελος;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Ναι, αλλά **αντιγράφεται το περιεχόμενο του φακέλου, όχι ο ίδιος ο φάκελος.**

</div>

Με `WORKDIR /app` και αρχεία `app/main.py`, `utils/helpers.py`, `config.toml`:

```dockerfile
COPY app utils config.toml ./
```
```
/app/main.py         ← ο φάκελος app/ "εξαφανίστηκε"
/app/helpers.py      ← ο φάκελος utils/ "εξαφανίστηκε"
/app/config.toml
```

Για να **κρατήσω τους φακέλους**, δίνω τον φάκελο στον προορισμό, με ξεχωριστό `COPY` για τον καθένα:
```dockerfile
COPY app ./app          # → /app/app/main.py
COPY utils ./utils      # → /app/utils/helpers.py
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Αν δύο πηγές έχουν αρχείο με το ίδιο όνομα, **το τελευταίο αντικαθιστά το προηγούμενο**.

</div>

| Εντολή (`WORKDIR /app`) | Αποτέλεσμα |
|---|---|
| `COPY main.py .` | `/app/main.py` |
| `COPY main.py ./src/` | `/app/src/main.py` |
| `COPY *.json ./` | Όλα τα `.json` στο `/app/` |
| `COPY ./app ./app` | Περιεχόμενο του `app/` στο `/app/app/` |
| `COPY ./app .` | Περιεχόμενο του `app/` απευθείας στο `/app/` |
| `COPY . .` | Όλο το build context στο `/app/` |

---

### <span style="color:#1E88E5">❓ Τι άλλο πρέπει να ξέρω για το `COPY`;</span>

- **Paths με κενά**: το κενό χωρίζει τα ορίσματα, οπότε χρησιμοποιώ μορφή JSON: `COPY ["my file.txt", "/app/"]`
- **<span style="color:#43A047">--from=&lt;image&gt;</span>**: αντιγράφει από άλλο image αντί από τον υπολογιστή μου.
- **<span style="color:#43A047">--chown=&lt;user&gt;:&lt;group&gt;</span>**: ορίζει ιδιοκτήτη των αρχείων (χρήσιμο με μη-root χρήστη).

| Εντολή | Τι κάνει | Πότε |
|---|---|---|
| **`COPY`** | Μόνο αντιγραφή | **Προεπιλογή**: προβλέψιμο |
| **`ADD`** | Αντιγραφή + κατέβασμα από URL + αποσυμπίεση `.tar` | Μόνο αν χρειάζομαι αυτές τις δυνατότητες |

## <span style="color:#1E88E5">🛠️ Εγκατάσταση uv</span>

### <span style="color:#1E88E5">❓ Πώς βάζω το uv στο image;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Το base image δεν έχει uv. Το uv είναι ένα μόνο εκτελέσιμο αρχείο, οπότε υπάρχουν δύο ισοδύναμοι τρόποι να το βάλω.

</div>

```dockerfile
# Επιλογή Α: αντιγραφή του εκτελέσιμου από το επίσημο image του uv
COPY --from=ghcr.io/astral-sh/uv:X.Y.Z /uv /uvx /bin/

# Επιλογή Β: εγκατάσταση μέσω pip
RUN pip install --no-cache-dir uv==X.Y.Z
```

| Επιλογή | Πλεονέκτημα | Πότε |
|---|---|---|
| **`COPY --from`** | Πιο γρήγορο και «καθαρό», δεν αγγίζει την Python του συστήματος, δουλεύει και χωρίς Python. Προτείνεται από την τεκμηρίωση του uv | Προεπιλογή |
| **`pip install`** | Πιο οικείο, πιο εύκολο να διαβαστεί | Αν το προτιμώ για απλότητα |

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Και στους δύο τρόπους **κλειδώνω την έκδοση**: `X.Y.Z` = η τοπική μου, από `uv --version`.

</div>

---

### <span style="color:#1E88E5">❓ Τι σημαίνει `/uv /uvx /bin/`;</span>

```
COPY --from=ghcr.io/astral-sh/uv:X.Y.Z   /uv   /uvx   /bin/
                    │                      │      │       │
                    │                      │      │       └── προορισμός: φάκελος στο ΔΙΚΟ ΜΟΥ image
                    │                      │      └── πηγή 2
                    │                      └── πηγή 1
                    └── από ποιο image παίρνω τα αρχεία
```

- Λόγω `--from`, οι πηγές είναι αρχεία **μέσα στο image του uv**, όπου βρίσκονται στη ρίζα (`/`).
- **<span style="color:#43A047">/bin/</span>**: φάκελος όπου το Linux ψάχνει εντολές (είναι στο `PATH`). Έτσι μετά γράφω απλά `uv sync`.
- **<span style="color:#43A047">uv</span>**: το κύριο εργαλείο. **<span style="color:#43A047">uvx</span>**: τρέχει εργαλεία χωρίς μόνιμη εγκατάσταση (σαν το `npx`). Αν δεν το χρειάζομαι, μπορεί να λείπει.

## <span style="color:#1E88E5">⚙️ Περιβάλλον και φάκελος εργασίας</span>

### <span style="color:#1E88E5">❓ Τι κάνουν οι μεταβλητές `ENV`;</span>

```dockerfile
ENV PYTHONDONTWRITEBYTECODE=1 \
    PYTHONUNBUFFERED=1
```

Το `ENV` ορίζει **μεταβλητές περιβάλλοντος**, που ισχύουν στο build και όταν τρέχει το container.
- **<span style="color:#43A047">PYTHONDONTWRITEBYTECODE=1</span>**: η Python δεν γράφει αρχεία `.pyc`, που σε container απλώς πιάνουν χώρο.
- **<span style="color:#43A047">PYTHONUNBUFFERED=1</span>**: τα logs βγαίνουν αμέσως αντί να κρατιούνται σε buffer.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Χωρίς `PYTHONUNBUFFERED=1`, συχνά το `docker logs` δεν δείχνει τίποτα και νομίζω ότι η εφαρμογή δεν τρέχει.

</div>

---

### <span style="color:#1E88E5">❓ Τι κάνει το `WORKDIR`;</span>

```dockerfile
WORKDIR /app
```

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Ορίζει τον **τρέχοντα φάκελο** για όλες τις επόμενες εντολές. Αν δεν υπάρχει, τον δημιουργεί. Κάθε σχετικό path που ακολουθεί (π.χ. `.`) σημαίνει `/app`.

</div>

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Όχι `RUN cd /app`. Κάθε `RUN` τρέχει ξεχωριστά, οπότε το `cd` δεν «κρατιέται» στην επόμενη εντολή.

</div>

## <span style="color:#1E88E5">📥 Εξαρτήσεις και κώδικας</span>

### <span style="color:#1E88E5">❓ Γιατί αντιγράφω πρώτα τις εξαρτήσεις και μετά τον κώδικα;</span>

```dockerfile
COPY pyproject.toml uv.lock ./
RUN uv sync --locked --no-install-project --no-dev --no-cache

COPY ./app ./app
RUN uv sync --locked --no-dev --no-cache
```

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 **Caching.** Κάθε εντολή δημιουργεί layer, και αν ένα layer αλλάξει, ξαναχτίζονται αυτό και όλα όσα ακολουθούν. Οι εξαρτήσεις αλλάζουν σπάνια και η εγκατάστασή τους είναι αργή. Ο κώδικας αλλάζει συνέχεια. Χωρίζοντάς τα, μια αλλαγή στον κώδικα ξανατρέχει μόνο τα δύο τελευταία (γρήγορα) βήματα.

</div>

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **Κανόνας:** ό,τι αλλάζει σπάνια → ψηλά στο Dockerfile. Ό,τι αλλάζει συχνά → χαμηλά.

</div>

- Το πρώτο `uv sync` εγκαθιστά **μόνο τις εξαρτήσεις**. Το δεύτερο, αφού μπει ο κώδικας, εγκαθιστά **το ίδιο το project**.
- Το `COPY ./app ./app` κρατά τον φάκελο `app`, ώστε να δουλεύει το `app.main:app` στο `CMD`.
- Το `RUN` εκτελεί εντολή **κατά το build**. Το `uv sync` δημιουργεί virtual environment στο `/app/.venv`.

---

### <span style="color:#1E88E5">❓ Τι σημαίνουν οι επιλογές του `uv sync`;</span>

- **<span style="color:#43A047">--locked</span>**: αν το `uv.lock` δεν ταιριάζει με το `pyproject.toml`, το build **αποτυγχάνει** αντί να εγκαταστήσει κάτι διαφορετικό.
- **<span style="color:#43A047">--no-install-project</span>**: μόνο οι εξαρτήσεις, όχι το project (ο κώδικας δεν έχει αντιγραφεί ακόμα).
- **<span style="color:#43A047">--no-dev</span>**: χωρίς dev dependencies (π.χ. `pytest`, `ruff`), που δεν χρειάζονται για να τρέξει η εφαρμογή.
- **<span style="color:#43A047">--no-cache</span>**: δεν κρατά cache με τα κατεβασμένα πακέτα → μικρότερο image.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ Αν αλλάξω εξαρτήσεις στο `pyproject.toml` με το χέρι, τρέχω `uv lock` τοπικά πριν το build. Αλλιώς το `--locked` σταματά το build.

</div>

## <span style="color:#1E88E5">🚀 Εκκίνηση της εφαρμογής</span>

### <span style="color:#1E88E5">❓ Γιατί χρειάζεται το `ENV PATH`;</span>

```dockerfile
ENV PATH="/app/.venv/bin:$PATH"
```

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Ο `uvicorn` εγκαταστάθηκε στο `/app/.venv/bin`. Βάζοντας αυτόν τον φάκελο **πρώτο** στο `PATH`, το `CMD` τον βρίσκει. Είναι ισοδύναμο με το `source .venv/bin/activate`.

</div>

| Επιλογή | Πλεονέκτημα | Πότε |
|---|---|---|
| **`ENV PATH` + `CMD ["uvicorn", ...]`** | Η εφαρμογή ξεκινά απευθείας | Προεπιλογή |
| **`CMD ["uv", "run", "uvicorn", ...]`** | Χωρίς `ENV PATH` | Για απλότητα. Όμως ελέγχει/συγχρονίζει το περιβάλλον σε κάθε εκκίνηση |

---

### <span style="color:#1E88E5">❓ Ανοίγει το `EXPOSE` τη θύρα;</span>

```dockerfile
EXPOSE 8000
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ **Όχι.** Είναι μόνο **τεκμηρίωση** («η εφαρμογή ακούει στη 8000»). Η θύρα ανοίγει προς τον host μόνο με `-p` στο `docker run`.

</div>

---

### <span style="color:#1E88E5">❓ Τι κάνει το `CMD`;</span>

```dockerfile
CMD ["uvicorn", "app.main:app", "--host", "0.0.0.0", "--port", "8000"]
```

<div style="background:rgba(142,36,170,0.12); border-left:5px solid #8E24AA; padding:10px 14px; border-radius:6px;">

🧠 **RUN** = κατά το χτίσιμο του image (εγκατάσταση). **CMD** = κατά την εκκίνηση του container (εκτέλεση).

</div>

- **<span style="color:#43A047">app.main:app</span>**: «στο module `app/main.py`, τρέξε το αντικείμενο `app`».
- **<span style="color:#43A047">--host 0.0.0.0</span>**: ακούει σε όλες τις διεπαφές δικτύου του container.
- **Μορφή JSON array**: η εντολή τρέχει απευθείας, χωρίς shell, οπότε ο uvicorn λαμβάνει το σήμα τερματισμού στο `docker stop` και κλείνει σωστά.

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ **Το πιο συχνό λάθος:** χωρίς `--host 0.0.0.0`, ο uvicorn ακούει στο `127.0.0.1`, δηλαδή στο localhost **του container**. Το container τρέχει, αλλά από τον browser δεν απαντά τίποτα. Επίσης, όχι `--reload` σε production.

</div>

## <span style="color:#1E88E5">🔒 Καθαριότητα και ασφάλεια</span>

### <span style="color:#1E88E5">❓ Τι βάζω στο `.dockerignore`;</span>

<div style="background:rgba(67,160,71,0.12); border-left:5px solid #43A047; padding:10px 14px; border-radius:6px;">

💡 Σαν το `.gitignore`: ό,τι είναι εκεί **δεν στέλνεται στο build context**, άρα δεν μπορεί να μπει στο image.

</div>

```
.venv/
__pycache__/
*.pyc
.git/
.env
```

<div style="background:rgba(251,140,0,0.12); border-left:5px solid #FB8C00; padding:10px 14px; border-radius:6px;">

⚠️ **`.venv/`**: το τοπικό venv είναι φτιαγμένο για το δικό μου λειτουργικό και δεν δουλεύει σε Linux container. **`.env`**: κωδικοί και secrets δεν μπαίνουν ποτέ σε image. Δίνονται κατά την εκτέλεση (`docker run --env-file .env`).

</div>

---

### <span style="color:#1E88E5">❓ Πρέπει η εφαρμογή να τρέχει ως root;</span>

Όχι, ιδανικά. Αν κάποιος εκμεταλλευτεί ένα κενό ασφαλείας, ως απλός χρήστης θα έχει λιγότερα δικαιώματα. Πριν το `CMD`:

```dockerfile
RUN useradd --create-home appuser   # δημιουργεί απλό χρήστη
USER appuser                        # η εφαρμογή τρέχει ως appuser
```

## <span style="color:#1E88E5">🧭 Σύνοψη</span>

### <span style="color:#1E88E5">❓ Σε ποια σημεία ενός Dockerfile έχω επιλογές;</span>

| Απόφαση | Επιλογές | Προεπιλογή | Πότε το άλλο |
|---|---|---|---|
| **Παραλλαγή base image** | `slim` / `alpine` / πλήρες | `slim` | `alpine` για μικρό μέγεθος, πλήρες για compile |
| **Πλήθος stages** | Ένα `FROM` / multi-stage | Ένα (Python, Node) | Multi-stage όταν υπάρχει βήμα build (Go, Java, frontend) |
| **Εγκατάσταση uv** | `COPY --from` / `pip install` | `COPY --from` | `pip` για απλότητα |
| **Εγκατάσταση εξαρτήσεων** | Ένα βήμα / δύο (πρώτα deps, μετά κώδικας) | Δύο βήματα (caching) | Ένα, μόνο σε πολύ μικρά/πρόχειρα projects |
| **Εκκίνηση εφαρμογής** | `PATH` στο `.venv` / `uv run` | `PATH` | `uv run` για απλότητα |
| **Αντιγραφή αρχείων** | `COPY` / `ADD` | `COPY` | `ADD` για URL ή `.tar` |
| **Χρήστης εκτέλεσης** | root / απλός χρήστης | Απλός χρήστης | root μόνο για πρόχειρα τοπικά τεστ |
